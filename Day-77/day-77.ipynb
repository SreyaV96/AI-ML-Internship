### Task 1 
Create five sample documents with metadata including: 
- Filename 
- Page number 
- Department 

```python
from langchain_core.documents import Document

documents = [
    Document(
        page_content="HR policy states that employees are entitled to 20 days of annual leave.",
        metadata={
            "filename": "hr_policy.pdf",
            "page": 1,
            "department": "HR"
        }
    ),

    Document(
        page_content="HR recruitment guidelines require at least two interview rounds.",
        metadata={
            "filename": "recruitment_manual.pdf",
            "page": 5,
            "department": "HR"
        }
    ),

    Document(
        page_content="The finance department prepares quarterly financial reports.",
        metadata={
            "filename": "finance_report.pdf",
            "page": 10,
            "department": "Finance"
        }
    ),

    Document(
        page_content="The IT department performs weekly server backups.",
        metadata={
            "filename": "it_manual.pdf",
            "page": 3,
            "department": "IT"
        }
    ),

    Document(
        page_content="Marketing campaigns are evaluated using customer engagement metrics.",
        metadata={
            "filename": "marketing_strategy.pdf",
            "page": 7,
            "department": "Marketing"
        }
    )
]
```

### Task 2 
Store the documents in ChromaDB with metadata. 

```python
from langchain_community.vectorstores import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
embedding = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

db = Chroma.from_documents(
    documents=documents,
    embedding=embedding,
    persist_directory="./company_db"
)

db.persist()

print("Documents stored successfully.")
```

### Task 3 
Retrieve only HR documents using metadata filtering. 

```python
results = db.similarity_search(
    query="leave policy",
    k=5,
    filter={"department": "HR"}
)

for doc in results:
    print("Filename:", doc.metadata["filename"])
    print("Page:", doc.metadata["page"])
    print("Department:", doc.metadata["department"])
    print("Content:", doc.page_content)
    print("-"*50)
```

### Task 4 
Ask questions and display the source file name and page number with each answer. 

```python
question = "How many annual leave days are employees entitled to?"

results = db.similarity_search(
    question,
    k=2
)

for doc in results:
    print("Answer:")
    print(doc.page_content)
    print("Source File:", doc.metadata["filename"])
    print("Page Number:", doc.metadata["page"])
    print("="*60)
```
