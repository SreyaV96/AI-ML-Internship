### Task 3 
Research and write short notes on: 
- FAISS 
- ChromaDB 
- Pinecone 

#### FAISS
FAISS (Facebook AI Similarity Search) is an open-source library developed by Meta for fast similarity search on vector embeddings. It is commonly used in RAG applications to retrieve relevant documents efficiently.

#### ChromaDB
ChromaDB is an open-source vector database designed for AI applications. It stores embeddings and metadata, making it easy to build chatbots and knowledge bases.

#### Pinecone
Pinecone is a cloud-based vector database that provides fast and scalable vector search. It is widely used for enterprise RAG applications without requiring infrastructure management.

### Task 5 
Explain how a chatbot can answer questions from a PDF using the RAG approach. 

1. Upload the PDF document.
2. Extract text from the PDF.
3. Convert the text into embeddings.
4. Store the embeddings in a vector database.
5. When a user asks a question:
   - Convert the question into an embedding.
   - Search the vector database.
   - Retrieve relevant text from the PDF.
   - Send the retrieved content and the question to the Large Language Model (LLM).
   - Generate and return the final response.
