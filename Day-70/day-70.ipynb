### Task 5 
Research two frameworks used for building RAG applications: 
- LangChain 
- LlamaIndex 
Write a short note (5–6 lines) on each. 

##### 1. LangChain

LangChain is one of the most popular frameworks for developing LLM-powered applications, especially RAG systems. It provides ready-made components for document loading, text chunking, embedding generation, vector database integration, prompt templates, and conversational memory. It supports multiple LLM providers such as OpenAI, Gemini, Anthropic, and Ollama, along with vector databases like ChromaDB, Pinecone, and FAISS. Its modular architecture makes it easy to build, test, and deploy RAG chatbots with minimal code.



##### 2. LlamaIndex

LlamaIndex is a framework designed specifically for connecting LLMs with external data sources. It simplifies document ingestion, indexing, retrieval, and query processing for RAG applications. LlamaIndex supports PDFs, databases, APIs, cloud storage, and many other data sources. It integrates well with vector databases such as ChromaDB and FAISS and works with popular LLMs including OpenAI, Gemini, and local models. Its focus on efficient indexing and retrieval makes it an excellent choice for building knowledge-based AI assistants.
