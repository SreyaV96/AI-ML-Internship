### Task 2 
Write a Python program to tokenize the sentence: 
Artificial Intelligence is changing the world. 

```python
from nltk.tokenize import word_tokenize
text = "Artificial Intelligence is changing the world"
tokens = word_tokenize(text)
print(tokens)
```

### Task 3 
Remove stop words from: 
I want to book a train ticket to Delhi. 

```python
from nltk.corpus import stopwords
text = "I want to book a train ticket to Delhi"
tokens = word_tokenize(text)

stop_words = set(stopwords.words("english"))

filtered = []
for word in tokens:
    if word.lower() not in stop_words:
        filtered.append(word)
print(filtered)
```

### Task 4 
Apply stemming to: 
- Playing 
- Running 
- Reading 
- Learning 

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
print(stemmer.stem("Playing"))
print(stemmer.stem("Running"))
print(stemmer.stem("Reading"))
print(stemmer.stem("Learning"))
```

### Task 5 
Build a chatbot that answers questions about: 
- Fees 
- Course Duration 
- Eligibility 
- Contact Number 
- College Location 
using NLTK tokenization. 

```python
print("====College Enquiry Bot====")
print("Type'Bye'to Exit")

while True:
    user = input("You:").lower()
    if user == "bye":
        print("Bot: Goodbye! Thank you for contacting us.")
        break
    tokens = word_tokenize(user)
    if "hello" in tokens or "hi" in tokens:
        print("Bot: Hello! How can I help you?") 
    elif "fee" in tokens or "fees" in tokens:
        print("Bot: The course fee is ₹25,000.")
    elif "duration" in tokens or "long" in tokens:
        print("Bot: Course duration is 3 months.") 
    elif "eligibility" in tokens or "qualification" in tokens:
        print("Bot: Any degree with at least 50% marks.")
    elif "contact" in tokens or "phone" in tokens or "number" in tokens or "call" in tokens:
        print("Bot: Contact us at 9876543210.")
    elif "location"in tokens or "address" in tokens or "where" in tokens:
        print("Bot: ABC College, Kochi, Kerala, 600111")
    else: 
        print("Bot: Sorry! I couldn't understand.") 

    

```
