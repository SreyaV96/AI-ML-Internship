```python
from langchain_core.prompts import PromptTemplate

template = """
You are an educational chatbot.

Subject:
{subject}

Topic:
{topic}

Student Level:
{level}

Instructions:
- Explain the topic in simple language.
- Give examples.
- End with 3 practice questions.

Answer:
"""

prompt = PromptTemplate(
    input_variables=["subject", "topic", "level"],
    template=template
)
```
