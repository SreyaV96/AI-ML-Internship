### Task 1 

Create three different prompt templates for: 
- Educational chatbot 
- Banking chatbot 
- Customer support chatbot 

```python
import langchain
print(langchain.__version__)
```

```python
#Educational chatbot
from langchain_core.prompts import PromptTemplate

template = """
You are an educational chatbot.

Subject:
{subject}

Topic:
{topic}

Student Level:
{level}

Instructions:
- Explain the topic in simple language.
- Give examples.
- End with 3 practice questions.

Answer:
"""

prompt = PromptTemplate(
    input_variables=["subject", "topic", "level"],
    template=template
)
```

```python
#Bank chatbot
from langchain_core.prompts import PromptTemplate

template = """
You are a banking chatbot.

Customer Query:
{query}

Instructions:
- Answer professionally.
- Explain banking terms in simple language.
- Never reveal confidential information.
- If account-specific information is required, advise the customer to contact the bank.
- Do not generate fake account details.

Response:
"""

banking_prompt = PromptTemplate(
    input_variables=["query"],
    template=template
)
```

```python
#Customer support chatbot
from langchain_core.prompts import PromptTemplate

template = """
You are a customer support chatbot.

Customer Issue:
{issue}

Instructions:
- Be polite and empathetic.
- Understand the customer's problem.
- Provide step-by-step troubleshooting.
- Suggest contacting support if the issue cannot be solved.
- End by asking whether further assistance is needed.

Response:
"""

support_prompt = PromptTemplate(
    input_variables=["issue"],
    template=template
)
```

### Task 3 

Design a prompt template that answers questions only from uploaded PDF 
documents. 

```python
from langchain_core.prompts import PromptTemplate

template = """
You are a document-based chatbot.

Question:
{question}

Instructions:
- Answer ONLY from the uploaded PDF.
- Do not use outside knowledge.
- If the answer is not available, say:
  "The uploaded PDF does not contain this information."
- Mention the page number whenever possible.
- Never make assumptions.

Answer:
"""

pdf_prompt = PromptTemplate(
    input_variables=["question"],
    template=template
)
```
