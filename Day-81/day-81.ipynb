## Theory 
1. Define a Capstone Project. 

A Capstone Project is a practical, comprehensive project undertaken at the end of a course, training program, or internship to demonstrate the knowledge and skills acquired during the learning process.
It involves identifying a real-world problem, designing a solution, implementing it using suitable technologies, testing its performance, and presenting the final results.

2. Explain the AI project development lifecycle
    1. **Problem Identification** 
    2. **Requirement Analysis**  
    4. **AI Model and Pipeline Development** 
    5. **Application Development**
    6. **Testing and Evaluation** 
    7. **Deployment**
    8. **Monitoring and Maintenance**

3. Differentiate functional and non-functional requirements.

| Functional Requirements | Non-Functional Requirements |
|---|---|
| Define what the system must do. | Define how well the system must operate. |
| Describe features and system behavior. | Describe quality attributes and operational constraints. |
| Usually verified through feature-based tests. | Usually verified through performance, security, usability, and reliability tests. |
| Example: Users can upload PDF files. | Example: API keys must be stored securely,The application should handle invalid files without crashing.|
