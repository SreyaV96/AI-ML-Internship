## Theory
#### 1. Define System Design

**System design** is the process of defining the structure, components, modules, interfaces, and data flow required to develop a software system. It explains how different parts of the system work together to meet the requirements.

System design generally includes high-level design and low-level design.

- **High-Level Design (HLD):** Describes the overall architecture, major components, and interactions between them.
- **Low-Level Design (LLD):** Describes the internal details of individual components, such as classes, functions, algorithms, and database structures.


#### 2. Define System Architecture
**System architecture** is the overall structural organization of a software system. It defines the major components, their responsibilities, and how they communicate with one another.

System architecture provides a high-level blueprint for developing and maintaining the application.


#### 3. Explain the Role of Frontend, Backend, Database, and AI Model

##### a. Frontend

The **frontend** is the part of an application that users interact with directly. It provides the user interface and displays information.

**Responsibilities:**
- Accept user input.
- Allow users to upload files.
- Send requests to the backend.
- Display answers and source references.

**Example:** An HTML page where users upload research papers and enter questions.

##### b. Backend

The **backend** handles application logic, processes requests, and connects the frontend with databases and AI services.

**Responsibilities:**
- Receive and validate requests.
- Process uploaded documents.
- Communicate with the database and AI model.
- Return responses to the frontend.

**Example:** FastAPI endpoints that handle PDF uploads and question-answering requests.

##### c. Database

A **database** stores and organizes information so that it can be retrieved and managed when needed.

**Responsibilities:**
- Store application data.
- Retrieve relevant information.
- Maintain document metadata.
- Support efficient searching.

**Example:** ChromaDB stores vector embeddings, text chunks, and source metadata for semantic retrieval in a RAG application.

### d. AI Model

An **AI model** is a computational model trained to perform tasks such as understanding text, recognizing patterns, making predictions, or generating content.

**Responsibilities:**
- Generate embeddings representing text.
- Support semantic search and retrieval.
- Generate responses based on provided context.
- Perform other AI-related tasks depending on the application.

**Example:** Sentence Transformers generate embeddings, while the Gemini API generates answers based on retrieved research-paper content.

#### 4. What Is an API?

**API (Application Programming Interface)** is a set of rules and protocols that allows different software applications or components to communicate with one another.

An API defines how requests are made, what data is exchanged, and what responses are returned.

**Example:** In a RAG chatbot, a FastAPI endpoint can accept a user's question, retrieve relevant passages from ChromaDB, send the context to the OpenAI API, and return the generated answer.

Common HTTP methods used in web APIs include:

- **GET:** Retrieve information.
- **POST:** Submit data or create a resource.
- **PUT:** Replace an existing resource.
- **PATCH:** Partially update a resource.
- **DELETE:** Delete a resource.

#### 5. Differentiate SRS and System Design

| SRS (Software Requirement Specification) | System Design |
|---|---|
| Describes what the software must do. | Describes how the software will be built. |
| Defines functional and non-functional requirements. | Defines architecture, components, interfaces, and data flow. |
| Focuses on user needs and system expectations. | Focuses on technical structure and implementation. |
| Prepared during requirement analysis. | Developed using the requirements as a basis. |
| Used by developers and testers to understand expected behavior. | Used by developers to guide implementation. |
| Example: The system shall allow users to upload PDF files. | Example: A FastAPI upload endpoint receives PDFs and passes them to a PDF-processing module. |
