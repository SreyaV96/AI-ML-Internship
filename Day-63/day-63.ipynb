### Task 4 
Write a Python program that predicts an intent using a simple dictionary lookup. 

```python
training_data = {"hi": "Greeting",
    "hello": "Greeting",
    "good morning": "Greeting",
    "good evening": "Greeting",

    "courses available in your college?": "Course Enquiry",
    "courses offered": "Course Enquiry",
    
    "how can i apply for a management quota seat": "Management Quota Admission",

    "when does admission start": "Admission Date",
    "admission start date?": "Admission Date",

    "which university is the college affiliated with": "University Name",

    "course fee for mca per semester?": "Course Fee Enquiry",
    "college location": "College Location",
    "college address?": "College Location",
  
    "is hostel accommodation available?": "Hostel Enquiry",

    "contact number":"Contact Enquiry",

    "what is the contact number for the admission office?": "Contact Enquiry",

    "thank you": "Thank You",
    "thankyou": "Thank You"
}


user = input("You:").lower()

for text,intent in training_data.items():
    if text in user:
        print("Predicted Intent:", intent) 
        break 
else: 
    print("Intent not recognized.") 
```
