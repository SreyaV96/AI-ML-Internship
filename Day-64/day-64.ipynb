### Task 3 
Modify the chatbot to continue the conversation until the user types bye. 

````python
from openai import OpenAI 
client = OpenAI(api_key="YOUR_API_KEY") 
print("  AI Chatbot") 
print("Type 'bye' to exit.\n") 

while True:
 
    user = input("You: ") 
    if user.lower() == "bye": 
        print("Bot: Goodbye!") 
        break 

    response = client.responses.create( 
        model="gpt-4.1-mini", 
        input=user 
    ) 

    print("Bot:", response.output_text) 
````

### Task 4 
Create a chatbot that answers questions related to: 
- Artificial Intelligence 
- Python 
- Machine Learning 
- Data Science 

````python
from openai import OpenAI

client = OpenAI(api_key="YOUR_API_KEY")

print("AI Learning Bot")
print("Type 'bye' to exit.\n")
print("Ask about AI, Machine Learning, Data Science, and Python.\n")

while True:

    user = input("You: ")

    if user.lower() == "bye":
        print("Bot: Goodbye!")
        break

    response = client.responses.create(
        model="gpt-4.1-mini",
        input=f"""
You are an educational chatbot.

Answer only questions related to:
- Artificial Intelligence
- Python
- Machine Learning
- Data Science

If the question is not related to these topics, reply:
"Sorry, I can answer only questions related to Artificial Intelligence, Python, Machine Learning, and Data Science."

Question: {user}
"""
    )
    print("Bot:", response.output_text)
````

````python
## Chatbot using Google Gemini API
from google import genai

client = genai.Client(api_key="YOUR_API_KEY")

print("AI Learning Bot")
print("Type 'bye' to exit.\n")
print("Ask about AI, Machine Learning, Data Science, and Python.\n")

while True:

    user = input("You: ")

    if user.lower() == "bye":
        print("Bot: Goodbye!")
        break

    prompt = f"""
You are an educational chatbot.

Answer only questions related to:
- Artificial Intelligence
- Python
- Machine Learning
- Data Science

If the question is not related to these topics, reply:
"Sorry, I can answer only questions related to Artificial Intelligence, Python, Machine Learning, and Data Science."

Question: {user}
"""

    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt
    )

    print("Bot:", response.text)
````
