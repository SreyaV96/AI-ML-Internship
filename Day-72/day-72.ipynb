### Task 1 
Create a ChromaDB collection and store at least 10 sample text documents. 

```python
import chromadb
client = chromadb.Client()
collection = client.create_collection(name="Internship_info")

documents = ["Internship fee is Rs. 5000.",
    "Placement assistance is available after successful completion.",
    "Students receive a certificate after completion.",
    "The internship duration is 3 months.",
    "The internship is completely offline.", 
    "Machine Learning is covered in the syllabus.",
    "students will get a chance to work on real world projects",
    "The internship starts on August 1.",
    "Deep Learning is included in the syllabus.",
    "One-to-one mentor is available"]

ids = [str(i) for i in range (len(documents))]
collection.add(documents=documents, ids=ids)
print ("document stored successfully")

   
```

### Task 2 
Perform similarity searches for the following questions: 
- What is the internship fee? 
- How long is the internship? 
- Is placement assistance available? 

Observe the retrieved results. 

```python
queries = ["What is the internship fee?",
    "How long is the internship?",
    "Is placement assistance available?"]
for q in queries:
    print("\nQuery:",q)
    results = collection.query(query_texts = [q], n_results =3)
    print(results["documents"])
```

# Practical Task 4

## Comparison of Similarity Metrics

| **Feature** | **Cosine Similarity** | **Euclidean Distance** | **Dot Product** |
|-------------|-----------------------|------------------------|-----------------|
| **Definition** | Measures the angle between two vectors. | Measures the straight-line distance between two vectors. | Multiplies corresponding vector elements and sums the results. |
| **Range** | **-1 to 1** (typically **0 to 1** for text embeddings) | **0 to ∞** | **-∞ to ∞** (no fixed bounds) |
| **Best Value** | **1** (maximum similarity) | **0** (minimum distance; vectors are identical) | **Higher value** (largest score indicates highest similarity) |
| **Speed** | Moderate | Fast | Very Fast |
| **Best For** | Text embeddings and semantic similarity | Dense spatial or numeric data | Normalized text embeddings and vector retrieval |
| **Used In** | RAG, semantic search, NLP, chatbots | Clustering, KNN, image processing | Vector search, neural networks, recommendation systems |
| **Advantage** | Ignores vector magnitude and focuses on direction | Captures absolute distance between vectors | Computationally efficient and scalable for large datasets |

### Task 5 
Write a Python program that accepts a user query and displays the Top 3 most
