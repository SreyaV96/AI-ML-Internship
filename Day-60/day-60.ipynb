### Task 1 
Create a College Enquiry Menu Chatbot with the following options: 
- Course Fee 
- Duration 
- Eligibility 
- Location 
- Contact 
- Exit 

```python
print("===== College Enquiry Bot =====")
print("1. Course Fee")
print("2. Duration")
print("3. Eligibility")
print("4. Location")
print("5. Contact")
print("6. Exit")

while True:
    choice = input("Enter your choice: ")

    if choice == "1":
        print("Course Fee: ₹41,000")
    elif choice == "2":
        print("Course Duration: 6 months")
    elif choice == "3":
        print("Eligibility: Any degree with at least 50% marks.")
    elif choice == "4":
        print("Location: ABC College, Kochi")
    elif choice == "5":
        print("Contact: 9876543210, abccollege@gmail.com")
    elif choice == "6":
        print("Exit: Thank you for visiting!")
        break
    else:
        print("Invalid choice! Please try again.")
```
