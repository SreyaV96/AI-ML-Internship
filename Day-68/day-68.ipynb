Generate embeddings for these sentences:

````python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer("all-MiniLM-L6-v2")
````

````python
sentences = [
    "Artificial Intelligence", 
    "Machine Learning", 
    "Python Programming" 
] 
emebeddings = model.encode(sentences)
print(emebeddings)
````
