````python
from openai import OpenAI 
client = OpenAI(api_key="YOUR_API_KEY") 

conversation = [] 

print("  AI Chatbot") 
print("Type 'bye' to exit.")  

while True: 

    user = input("You: ") 

    if user.lower() == "bye": 
        print("Bot: Goodbye!") 
        break 

    conversation.append({ 
        "role": "user", 
        "content": user 
    }) 

 
    response = client.responses.create( 
        model="gpt-4.1-mini", 
        input=conversation 
    ) 

    answer = response.output_text 
    print("Bot:", answer) 


    conversation.append({ 
        "role": "assistant", 
        "content": answer 
    }) 
````
