```python
import chromadb
client = chromadb.Client()

collection = client.create_collection(name="college_data")
```

### Task 2 
Insert five sample documents into the collection.
 
Example: 
- AI Fee 
- Duration 
- Eligibility 
- Hostel 
- Placement 

```python
documents = [
    "AI course fee is Rs. 50,000.",
    "The internship duration is 6 months.",
    "Eligibility: Any graduate can apply.",
    "Hostel facilities are available.",
    "Placement assistance is provided."
]
ids = ["1", "2", "3", "4", "5"]
collection.add(documents=documents, ids=ids)
```

### Task 3 
Search for: 

How long is the internship? 

Observe the retrieved result. 

```python
results = collection.query(
    query_texts=["How long is the internship?"],
    n_results=1
)

print(results)
```

FAISS organizes vectors using different indexing methods to make similarity searches faster.

| Index Type | Purpose |
|------------|---------|
| IndexFlatL2 | Exact search using Euclidean distance |
| IndexFlatIP | Exact search using inner product |
| IVF | Faster search for large datasets |
| HNSW | Graph-based approximate search |
| PQ | Compresses vectors to save memory |
