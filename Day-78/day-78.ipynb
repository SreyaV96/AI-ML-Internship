### Task 1 
Create a chatbot using ConversationBufferMemory.

```python
import langchain
print(langchain.__version__)
```

```python
from langchain_ollama import ChatOllama
from langchain_classic.memory import ConversationBufferMemory
from langchain_classic.chains import ConversationChain

llm = ChatOllama(
    model="llama3.2"
)

memory = ConversationBufferMemory()

chatbot = ConversationChain(
    llm=llm,
    memory=memory,
    verbose=False
)

print("AI Chatbot")
print("Type 'bye' to exit.\n")

while True:
    user = input("You: ")

    if user.lower() == "bye":
        print("Bot: Goodbye!")
        break

    response = chatbot.predict(input=user)
    print("Bot:", response)
```

### Task 5 
Compare the following memory types: 
- Buffer Memory 
- Window Memory 
- Summary Memory 
- Entity Memory 

Include their advantages, disadvantages, and suitable use cases. 

 

| Memory Type        | Advantages                                                       | Disadvantages                               |Use Case                                       |
| ------------------ | ---------------------------------------------------------------- | ------------------------------------------- | ------------------------------------------------------- |
| **Buffer Memory**  | Remembers the entire conversation.                               | Uses more memory and tokens for long chats. | Short conversations, customer support.                  |
| **Window Memory**  | Stores only the last *n* messages, saving memory.                | Forgets older messages.                     | Long conversations where only recent context is needed. |
| **Summary Memory** | Summarizes old conversations to save tokens.                     | May miss small details.                     | Long chats and virtual assistants.                      |
| **Entity Memory**  | Remembers important details like names, places, and professions. | Doesn't remember the full conversation.     | Personalized chatbots and CRM systems.                  |
