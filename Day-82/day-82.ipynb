## Theory
##### 1. Define SRS. 
SRS (Software Requirement Specification) is a formal document that describes the functional and non-functional requirements of a software system. It specifies what the system should do, how it should perform, and the constraints under which it must operate.
An SRS serves as a reference for developers, testers, project managers, and stakeholders throughout the software development lifecycle.

#### 2. Explain Why SRS Is Important

An SRS is important for the following reasons:

1. **Clear Requirements:** It defines the features and expected behavior of the software.
2. **Common Understanding:** It ensures that developers, testers, and stakeholders understand the project requirements.
3. **Project Planning:** It helps estimate the time, cost, resources, and effort required.
4. **Reduced Errors:** It minimizes misunderstandings and reduces costly changes later in development.
5. **Testing:** It provides a basis for preparing test cases and verifying that requirements are satisfied.
6. **Project Scope Control:** It helps prevent unnecessary features from being added without approval.
7. **Documentation:** It serves as a reference for future maintenance and enhancements.

#### 4. Explain Project Scope

**Project scope** defines the boundaries of a project, including its objectives, features, deliverables, limitations, and activities. It clarifies what is included in the project and what is excluded.

Project scope helps the development team focus on the agreed requirements and avoid unnecessary work.
A clearly defined project scope helps manage time, resources, and stakeholder expectations.


#### 5. What Is a Use Case?

A **use case** describes how a user or another external system interacts with a software application to achieve a specific goal.

It identifies the actor, the goal, and the sequence of interactions between the actor and the system.
